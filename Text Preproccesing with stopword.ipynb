```python
paragraph = """Respected Principal, teachers, and my dear friends, today I feel privileged to share a speech on the remarkable life of Dr. APJ Abdul Kalam.
His full name is Avul Pakir Jainulabdeen Abdul Kalam, and he is one of the most loved Indian personalities.
Born in a humble family at Rameswaram, Dr. Kalam rose to become a renowned scientist and the President of India. His journey from selling newspapers as a boy to leading the country is a beacon for every dreamer.
He played a crucial role in the development of India’s missile and nuclear programs, earning the nickname ‘Missile Man of India’.
Dr. Kalam always motivated students with his talks, urging them to think creatively and never give up.
In every Apj Abdul Kalam Speech, he reminded us that failure is the stepping stone to success and that each of us can contribute to making India a developed nation.
He was admired for his humility, simplicity, and vision, and received numerous awards including the Padma Bhushan, Padma Vibhushan, and Bharat Ratna – the highest civilian award in India.
APJ Abdul Kalam also delivered an inspiring speech in the European Parliament, representing the dignity and intelligence of India on a global stage.
Let us always remember Abdul Kalam quotes for success such as “Dream, dream, dream. Dreams transform into thoughts and thoughts result in action.”
By applying his teachings in life, we can achieve our ambitions and uplift society. Thank you!"""

```

```python
from nltk.stem import PorterStemmer
stemming = PorterStemmer()
```

```python
import nltk
nltk.download('punkt_tab')
sentences = nltk.sent_tokenize(paragraph)
type (sentences)
```

```python
# Apply stopwords and filter and then apply stemming
for i in range(len(sentences)):
  words = nltk.word_tokenize(sentences[i])
  words = [stemming.stem(word) for word in words if word not in set (stopwords.words("english"))]
  sentences [i] = ' '.join(words) # converting all words into  sentence
```
